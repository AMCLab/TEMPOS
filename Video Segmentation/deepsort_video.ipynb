#### Convert the video frames into images to show in paper

```python
import cv2
import os

def save_video_frames(video_path, output_dir):
    """
    Save each frame of a video as individual image files in the specified directory
    
    Args:
        video_path (str): Path to the input video file
        output_dir (str): Directory to save the frames
    """
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Open the video file
    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        raise ValueError(f"Error opening video file: {video_path}")
    
    frame_count = 0
    success = True
    
    while success:
        # Read next frame
        success, frame = cap.read()
        
        if not success:
            break
        
        # Save frame as JPEG file
        frame_file = os.path.join(output_dir, f"frame_{frame_count:06d}.jpg")
        cv2.imwrite(frame_file, frame)
        
        frame_count += 1
    
    # Release resources
    cap.release()
    return frame_count

if __name__ == "__main__":

    video_path = "segmented_VIDEO1_renset50.mp4"  
    output_directory = "segmented_VIDEO1_renset50"
    
    print(f"Extracting frames from {video_path}...")
    total_frames = save_video_frames(video_path, output_directory)
    print(f"Successfully saved {total_frames} frames to {output_directory}/")
```
