Datasette needs a .csv file to be launched. To create a .csv file for a directory with .dm images:

````python
import os
import ncempy.io.dm as dm
import pandas as pd
````

````python
# Path to a directory with your .dm files
path = './dm4_example'
````

````python
myData = list()

files = os.listdir(path) # list the image files in the path

for image in files: # loop through the image files

    load_in = os.path.join(path, image) # path to the image

    if (image.endswith(".dm3") or image.endswith(".dm4")):
                
        dm4_file = dm.dmReader(load_in) #read the dm3 file

        image_name = os.path.splitext(image)[0]
        
        dm4_file.pop('data')

        dm4_file['TEM image'] = str(path + '_' + image_name +'.png')
        dm4_file['coords'] = dm4_file['coords'][0]
        dm4_file['pixelSize'] = str(dm4_file['pixelSize'][0])
        dm4_file['pixelUnit'] = str(dm4_file['pixelUnit'][0])
        dm4_file['file_path'] = path

        myData.append(dm4_file)

df = pd.DataFrame(myData)

try:

    os.mkdir(path + '/datasette/')

except: pass

df.to_csv(path + '/datasette/dm4_images.csv', index=False, header=True)  

print("CSV saved at: " + path + '/datasette/dm4_images.csv')
````

#### You can create and launch a datasette application following these steps in your terminal:

Ps: if you already have the .db file, just install the required packages and type ```datasette dm4_images.db``` on terminal

If you haven't yet done so, 

```
pip install datasette
pip install sqlite-utils
```
Next, create a .db file using a .csv file. This can be the any .csv, e.g. generated above or generated from the segment_images.ipynb:

dm4_images is the name of your datasette and dm4_images.csv is the path to your .csv

```
sqlite-utils insert dm4_images.db dm4_images dm4_images.csv --csv
```

Optional: you can extact columns from the .db file to make data searchable. "pixelUnit", "pixelSize", "resolution (µm)" and "file_path" are examples of
column names. Change them accordingly.

```
sqlite-utils extract dm4_images.db  dm4_images   pixelUnit
sqlite-utils extract dm4_images.db  dm4_images   pixelSize
sqlite-utils extract dm4_images.db  dm4_images   'resolution (µm)'
sqlite-utils extract dm4_images.db  dm4_images    file_path
```

Finally, launch your datasette application:

```
datasette dm4_images.db 
```

Explore [Datasette Documentation](https://docs.datasette.io/en/stable/) and [Tutorials](https://datasette.io/tutorials/explore) for more tools and use cases.

##### To create a .db file from a CSV

````python

import sqlite3
import pandas as pd

# Read the CSV into a DataFrame
df = pd.read_csv("./dm4_example/datasette/dm4_images.csv")

# Connect to the SQLite database
conn = sqlite3.connect("./dm4_example/datasette/dm4_images.db")

# Insert the data into the database
df.to_sql('dm4_images', conn, if_exists='replace', index=False)

# Close the connection
conn.close()
````
